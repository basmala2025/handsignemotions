```python
import cv2
import mediapipe as mp
import numpy as np
import joblib

# === Load Model + Scaler + Label Encoder ===
model = joblib.load("hand_sign_model.pkl")
scaler = joblib.load("scaler.pkl")
label_encoder = joblib.load("label_encoder.pkl")

# === MediaPipe Setup ===
mp_hands = mp.solutions.hands
mp_draw = mp.solutions.drawing_utils
hands = mp_hands.Hands(max_num_hands=1, min_detection_confidence=0.6, min_tracking_confidence=0.6)

# === Start Camera ===
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    frame = cv2.flip(frame, 1)
    rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    results = hands.process(rgb)

    landmarks = []

    if results.multi_hand_landmarks:
        hand = results.multi_hand_landmarks[0]
        mp_draw.draw_landmarks(frame, hand, mp_hands.HAND_CONNECTIONS)

        for lm in hand.landmark:
            landmarks.extend([lm.x, lm.y, lm.z])

        # === Convert to Model Input Shape ===
        X_input = np.array(landmarks).reshape(1, -1)

        # === Apply Same Scaling from Training ===
        X_input = scaler.transform(X_input)

        # === Predict ===
        pred = model.predict(X_input)[0]
        sign = label_encoder.inverse_transform([pred])[0]

        cv2.putText(frame, sign, (30, 80), cv2.FONT_HERSHEY_SIMPLEX, 1.8, (0,255,0), 4)

    cv2.imshow("REAL-TIME HAND SIGN", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
